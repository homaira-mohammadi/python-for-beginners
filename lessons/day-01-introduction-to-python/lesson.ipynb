## 1. What is Programming?

**Programming** means giving a computer **clear, step-by-step instructions** to do a task.

Think about asking a coffee machine for coffee. "Make me coffee" is not enough for a machine —
it needs every step, **in the right order**:

1. Heat water ☕
2. Add coffee
3. Add milk
4. Serve

If the steps are in the wrong order (serve *before* heating the water!), the result is wrong.
Computers work the same way: they do **exactly** what we tell them, in **exactly** that order.

Here are those steps written as a Python program:

````python
print("Step 1: Heat water")
print("Step 2: Add coffee")
print("Step 3: Add milk")
print("Step 4: Serve")
````

Notice that Python ran the lines **from top to bottom**, one after another. This is called **sequential** execution.

## 5. How Python Runs Your Code

What happens when you press *Run*?

| Step | What happens |
|---|---|
| 1. **Write code** | You type Python code in an editor, e.g. `print("Hello World")` |
| 2. **Interpreter reads** | The Python *interpreter* reads your code line by line and checks for syntax (grammar) errors |
| 3. **Compilation** | Python turns your code into **bytecode** — an in-between format the computer can process |
| 4. **Execution** | The **Python Virtual Machine (PVM)** runs the bytecode |
| 5. **Output** | The result appears on the screen: `Hello World` |

> You don't need to remember every step — just know that Python reads your code **from top to bottom** and stops if it finds a mistake.

## 6. Understanding `print()`

`print()` can display different kinds of data:

````python
print("Welcome to Python")   # text (always inside quotes)
print("My name is Homaira")  # text
print(2026)                  # a number (no quotes needed)
````

### Text needs quotes — numbers don't

Text in Python is called a **string**. A string must be inside quotes. You can use **double** `" "` or **single** `' '` quotes — just be consistent.

````python
print("I love Python")
print('I love Python too')
````

### Printing several things at once

Put a **comma** between items to print them on one line. Python adds a space between them automatically.

````python
print("I am", 20, "years old")
print("Python", "is", "fun")
````

### An empty line

`print()` with nothing inside prints a blank line — useful for making output easier to read.

````python
print("Line one")
print()
print("Line three (after a blank line)")
````

## 7. Printing Calculations

Python can work as a calculator. Write the calculation **without quotes**:

````python
print(5 + 3)    # addition
print(10 - 4)   # subtraction
print(10 * 2)   # multiplication
print(20 / 4)   # division
````

> 💡 Did you notice that `20 / 4` gives **`5.0`**, not `5`? Division with `/` always gives a **decimal number**.

### Quotes change everything!

Compare these two lines carefully:

````python
print("5 + 3")   # with quotes: this is TEXT, printed exactly as written
print(5 + 3)     # without quotes: this is MATH, Python calculates it
````

You can also mix text and a calculation with a comma:

````python
print("5 + 3 =", 5 + 3)
print("Minutes in a day:", 24 * 60)
````

## 8. Comments in Python

A **comment** is a note for humans. It starts with `#`. Python **ignores** everything after the `#` on that line.

Comments explain *what* your code does and *why* — for your teacher, your classmates, and for *you* next week!

````python
# This program prints my name
print("Ali")

print("Hello")  # a comment can also go at the end of a line

# print("This line will NOT run, because it is a comment")
````

> 💡 **Tip:** Programmers often "comment out" a line (put `#` in front of it) to switch it off temporarily without deleting it.

## 9. Python Syntax Rules

**Syntax** = the grammar rules of a programming language.

### Rule 1: Python is case-sensitive
Capital and small letters are **different**. `name`, `Name` and `NAME` are three different things.
And the command is `print` — **not** `Print` or `PRINT`.

### Rule 2: Indentation matters
Python uses **spaces at the start of a line** (indentation) to group code together.
You will use this a lot from Day 3. Here is a sneak peek:

````python
if True:
    print("Hello")   # these 4 spaces show this line belongs to the 'if'
````

## 10. Common Beginner Errors 🐛

Errors are **normal** — every programmer sees them every day! The trick is to **read the error message**.
The **last line** of the message tells you the type of error.

Run each cell below to see the error, then look at the fix.

### Error 1: Wrong capital letter

````python
Print("Hello")
````

`NameError: name 'Print' is not defined` → Python doesn't know `Print`. ✅ **Fix:** use lowercase `print("Hello")`.

### Error 2: Missing quotes

````python
print(Hello)
````

Without quotes, Python thinks `Hello` is the name of something, not text. ✅ **Fix:** `print("Hello")`.

### Error 3: Unclosed quotes

````python
print("Hello)
````

`SyntaxError: unterminated string literal` → every opening quote needs a closing quote. ✅ **Fix:** `print("Hello")`.

### Error 4: Brackets that don't match

````python
print("Hello"))
````

`SyntaxError: unmatched ')'` → there is one bracket too many. Every `(` needs exactly one matching `)`. ✅ **Fix:** `print("Hello")`.

> 🧠 **Error-reading checklist:** 1) Read the **last line** of the error. 2) Look at the **line number**. 3) Check spelling, capital letters, quotes and brackets.
