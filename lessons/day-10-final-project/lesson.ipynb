## 2. Handling Errors with `try` / `except` 🛡️

What happens if the user types `twenty` when we expect a number?

````python
age = int("twenty")
````

The program **crashes** with a `ValueError`. With `try` / `except` we can **catch** the error and handle it nicely:

```python
try:
    # code that might fail
except ValueError:
    # what to do if it fails
```

````python
text = "twenty"
try:
    age = int(text)
    print("Your age is", age)
except ValueError:
    print("❌ That's not a whole number!")
````

### A reusable "safe number" function
Combine a `while True` loop, `try` / `except` and a function — you'll use this in almost every project:

````python
def get_int(prompt, low=None, high=None):
    """Keep asking until the user types a whole number (optionally between low and high)."""
    while True:
        try:
            value = int(input(prompt))
        except ValueError:
            print("❌ Please type a whole number.")
            continue
        if (low is not None and value < low) or (high is not None and value > high):
            print(f"❌ Please type a number from {low} to {high}.")
            continue
        return value

age = get_int("Enter your age: ", 0, 120)
print("Thanks! You are", age)
````

Other common errors you can catch: `ZeroDivisionError` (dividing by 0), `KeyError` (missing dictionary key), `IndexError` (bad list index).

````python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("You can't divide by zero!")
````

## 3. How to Plan a Project 📝

Professional programmers don't start typing straight away. They follow steps:

1. **Understand** — what should the program do? Who uses it?
2. **Plan the data** — which variables, lists and dictionaries do you need?
3. **Break it into small jobs** — each job becomes a **function**
4. **Write pseudocode** — the steps in plain words
5. **Build one small piece at a time** and **test** it
6. **Improve** — handle bad input, make the output pretty

### Our project: Student Grade Manager 🎓

**What it does:** a teacher can add students, record their scores, see a report card and see class statistics.

**Data plan:** a dictionary — student name → list of scores:
```python
students = {
    "Sara": [88, 95],
    "Omar": [61],
}
```

**Jobs (functions):**
| Function | Job |
|---|---|
| `get_grade(avg)` | turn an average into a letter grade |
| `average(scores)` | average of a list (0 if empty) |
| `add_student(students)` | add a new student |
| `add_score(students)` | add a score for a student |
| `show_report(students)` | table of every student, average and grade |
| `show_stats(students)` | class average and top student |
| `main()` | the menu loop |

**Pseudocode for `main()`:**
```
repeat forever:
    show the menu
    ask for a choice
    if 1 → add student
    if 2 → add score
    if 3 → show report
    if 4 → show statistics
    if 5 → say goodbye and stop
    otherwise → "invalid choice"
```

## 4. Build It Step by Step 🧱

### Step 1 — Helper functions (and test them!)

````python
def get_grade(avg):
    if avg >= 80:
        return "A"
    elif avg >= 70:
        return "B"
    elif avg >= 60:
        return "C"
    elif avg >= 50:
        return "D"
    return "F"

def average(scores):
    if len(scores) == 0:
        return 0
    return sum(scores) / len(scores)

# Quick tests ✅
print(get_grade(85), get_grade(72), get_grade(40))
print(average([80, 90]), average([]))
````

### Step 2 — Adding students and scores

````python
def add_student(students):
    name = input("Student name: ").strip().title()
    if name == "":
        print("❌ Name can't be empty.")
    elif name in students:
        print(f"❌ {name} is already registered.")
    else:
        students[name] = []
        print(f"✅ Added {name}.")

def get_score():
    while True:
        try:
            score = float(input("Score (0-100): "))
        except ValueError:
            print("❌ Please type a number.")
            continue
        if 0 <= score <= 100:
            return score
        print("❌ The score must be between 0 and 100.")

def add_score(students):
    name = input("Student name: ").strip().title()
    if name not in students:
        print(f"❌ {name} is not registered. Add them first.")
        return
    score = get_score()
    students[name].append(score)
    print(f"✅ Saved {score:g} for {name}.")
````

### Step 3 — Reports and statistics

````python
def show_report(students):
    if not students:
        print("No students yet.")
        return
    print(f"\n{'Name':<10}{'Scores':<20}{'Average':>8}  Grade")
    print("-" * 46)
    for name, scores in students.items():
        avg = average(scores)
        score_text = ", ".join(f"{s:g}" for s in scores) or "-"
        print(f"{name:<10}{score_text:<20}{avg:>8.1f}  {get_grade(avg)}")

def show_stats(students):
    with_scores = {name: average(s) for name, s in students.items() if s}
    if not with_scores:
        print("No scores yet.")
        return
    class_avg = sum(with_scores.values()) / len(with_scores)
    top = max(with_scores, key=with_scores.get)
    print(f"Class average: {class_avg:.1f}")
    print(f"Top student: {top} ({with_scores[top]:.1f}) 🏆")
````

> 💡 Two new shortcuts appeared above:
> - `{name: average(s) for name, s in students.items() if s}` is a **dictionary comprehension** — a one-line loop that builds a dictionary.
> - `max(with_scores, key=with_scores.get)` finds the key with the **biggest value**.
>
> You could write both with normal `for` loops too — try it!

### Step 4 — The menu (`main`) and run it! 🚀

````python
def main():
    students = {}
    while True:
        print("\n===== 🎓 STUDENT GRADE MANAGER =====")
        print("1. Add student")
        print("2. Add score")
        print("3. Show report")
        print("4. Show statistics")
        print("5. Quit")
        choice = input("Choose an option (1-5): ")

        if choice == "1":
            add_student(students)
        elif choice == "2":
            add_score(students)
        elif choice == "3":
            show_report(students)
        elif choice == "4":
            show_stats(students)
        elif choice == "5":
            print("Goodbye! 👋")
            break
        else:
            print("❌ Invalid choice, try again.")

main()
````
