## Project 1 ⭐⭐ — Rock, Paper, Scissors: Best of 3 ✊✋✌️

**Requirements:**
- The player plays against the computer (`random.choice`)
- Accept `rock`, `paper` or `scissors` in any capital letters; re-ask on invalid input
- The first to **2 wins** is the champion (ties don't count)
- Show the score after every round
- Use at least two functions, e.g. `get_player_choice()` and `winner(player, computer)`

```python
import random

OPTIONS = ["rock", "paper", "scissors"]
BEATS = {"rock": "scissors", "paper": "rock", "scissors": "paper"}  # key beats value

def get_player_choice():
    while True:
        choice = input("Rock, paper or scissors? ").strip().lower()
        if choice in OPTIONS:
            return choice
        print("❌ Please type rock, paper or scissors.")

def winner(player, computer):
    if player == computer:
        return "tie"
    if BEATS[player] == computer:
        return "player"
    return "computer"

player_wins = 0
computer_wins = 0
round_number = 1

while player_wins < 2 and computer_wins < 2:
    print(f"\n--- Round {round_number} ---")
    player = get_player_choice()
    computer = random.choice(OPTIONS)
    print("Computer chose", computer)

    result = winner(player, computer)
    if result == "player":
        player_wins += 1
        print("You win this round! 🎉")
    elif result == "computer":
        computer_wins += 1
        print("Computer wins this round! 🤖")
    else:
        print("It's a tie! 🤝")

    print(f"Score → You {player_wins} : {computer_wins} Computer")
    round_number += 1

if player_wins == 2:
    print("\n🏆 You are the champion!")
else:
    print("\n🤖 The computer is the champion. Try again!")
```

## Project 2 ⭐⭐ — Expense Tracker 💸

**Requirements:**
- A menu: 1. Add expense · 2. Show all · 3. Summary by category · 4. Quit
- Each expense is a dictionary: `{"category": "food", "amount": 12.5}` stored in a list
- The amount must be a positive number — use `try` / `except` and re-ask on bad input
- The summary shows the total per category and the grand total

```python
def get_amount():
    while True:
        try:
            amount = float(input("Amount (RM): "))
        except ValueError:
            print("❌ Please type a number.")
            continue
        if amount > 0:
            return amount
        print("❌ The amount must be more than 0.")

def add_expense(expenses):
    category = input("Category: ").strip().lower()
    amount = get_amount()
    expenses.append({"category": category, "amount": amount})
    print(f"✅ Added RM {amount:.2f} for {category}")

def show_all(expenses):
    if not expenses:
        print("No expenses yet.")
    for number, e in enumerate(expenses, start=1):
        print(f"{number}. {e['category']:<12} RM {e['amount']:>7.2f}")

def summary(expenses):
    totals = {}
    for e in expenses:
        totals[e["category"]] = totals.get(e["category"], 0) + e["amount"]
    for category, total in totals.items():
        print(f"{category:<12} RM {total:>7.2f}")
    print(f"{'TOTAL':<12} RM {sum(totals.values()):>7.2f}")

expenses = []
while True:
    print("\n1. Add expense  2. Show all  3. Summary  4. Quit")
    choice = input("Expense menu choice: ")
    if choice == "1":
        add_expense(expenses)
    elif choice == "2":
        show_all(expenses)
    elif choice == "3":
        summary(expenses)
    elif choice == "4":
        print("Goodbye! 👋")
        break
    else:
        print("❌ Invalid choice")
```

## Project 3 ⭐⭐⭐ — Hangman 🔤

**Requirements:**
- The computer picks a random word from a list
- Show the word as underscores, e.g. `_ _ t _ _ n`
- The player guesses one letter at a time; they have **6 lives**
- Don't accept guesses that are not a single letter, or letters already guessed
- The game ends when the word is complete (win) or lives reach 0 (lose)

```python
import random

WORDS = ["python", "variable", "function", "loop", "dictionary", "string"]

def show_word(word, guessed):
    display = ""
    for letter in word:
        if letter in guessed:
            display += letter + " "
        else:
            display += "_ "
    return display.strip()

def is_complete(word, guessed):
    for letter in word:
        if letter not in guessed:
            return False
    return True

word = random.choice(WORDS)
guessed = []
lives = 6

print("Welcome to Hangman! 🎯")
while lives > 0 and not is_complete(word, guessed):
    print(f"\n{show_word(word, guessed)}    Lives: {'❤️' * lives}")
    letter = input("Guess a letter: ").strip().lower()

    if len(letter) != 1 or not letter.isalpha():
        print("❌ Please type ONE letter.")
    elif letter in guessed:
        print("⚠️ You already guessed that letter.")
    elif letter in word:
        guessed.append(letter)
        print("✅ Good guess!")
    else:
        guessed.append(letter)
        lives -= 1
        print("❌ Not in the word.")

if is_complete(word, guessed):
    print(f"\n🎉 You won! The word was '{word}'.")
else:
    print(f"\n💀 Game over! The word was '{word}'.")
```
