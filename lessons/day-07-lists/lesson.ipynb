## 1. Why Do We Need Lists?

Imagine storing the names of 5 students:
```python
student1 = "Ali"
student2 = "Sara"
student3 = "Ahmad"
student4 = "Mariam"
student5 = "Omar"
```
What if there were 100 students? 😱 A **list** stores **many values in one variable**:

````python
students = ["Ali", "Sara", "Ahmad", "Mariam", "Omar"]
print(students)
````

A list:
- is written with **square brackets** `[ ]`
- has items separated by **commas**
- keeps items **in order**
- can be **changed** (unlike strings!)
- can hold any type — even mixed types

````python
scores = [85, 92, 78, 90]
prices = [1.5, 3.25, 4.0]
mixed = ["Ali", 20, 3.8, True]
empty = []

print(scores)
print(mixed)
print(len(scores))   # how many items?
print(type(scores))
````

## 2. Getting Items: Indexing and Slicing

Lists use indexes **exactly like strings** — starting at 0:
```
 "Ali"  "Sara"  "Ahmad"  "Mariam"  "Omar"
   0       1       2        3         4
  -5      -4      -3       -2        -1
```

````python
students = ["Ali", "Sara", "Ahmad", "Mariam", "Omar"]
print(students[0])     # first
print(students[2])
print(students[-1])    # last
print(students[1:3])   # slice: ['Sara', 'Ahmad']
print(students[:2])    # first two
````

## 3. Changing a List ✏️

### Change an item

````python
fruits = ["apple", "banana", "cherry"]
fruits[1] = "blueberry"
print(fruits)
````

### Add items — `append()` and `insert()`

````python
fruits = ["apple", "banana"]
fruits.append("cherry")        # add to the END
print(fruits)
fruits.insert(0, "mango")      # add at position 0
print(fruits)
````

### Remove items — `remove()`, `pop()` and `del`

````python
fruits = ["apple", "banana", "cherry", "mango"]

fruits.remove("banana")   # remove by VALUE
print(fruits)

last = fruits.pop()       # remove the LAST item and give it back
print("Popped:", last)
print(fruits)

del fruits[0]             # remove by INDEX
print(fruits)
````

> ⚠️ `remove()` gives an error if the item isn't in the list. Check with `in` first!

## 4. Checking with `in`

````python
shopping = ["eggs", "milk", "bread"]
print("milk" in shopping)      # True
print("cheese" in shopping)    # False

if "cheese" not in shopping:
    shopping.append("cheese")
print(shopping)
````

## 5. Looping Through a List 🔁

A `for` loop goes through **every item** — no index needed:

````python
students = ["Ali", "Sara", "Ahmad"]
for student in students:
    print("Hello,", student)
````

### Need the position too? Use `enumerate()`

````python
students = ["Ali", "Sara", "Ahmad"]
for number, student in enumerate(students, start=1):
    print(f"{number}. {student}")
````

### Building a new list in a loop

````python
numbers = [3, 8, 1, 6, 7, 2]
evens = []
for n in numbers:
    if n % 2 == 0:
        evens.append(n)
print("Even numbers:", evens)
````

## 6. Useful List Functions 🧰

| Function / method | What it does |
|---|---|
| `len(list)` | number of items |
| `sum(list)` | adds all numbers |
| `min(list)` / `max(list)` | smallest / largest |
| `sorted(list)` | gives a **new** sorted list |
| `list.sort()` | sorts the list **itself** |
| `list.reverse()` | reverses the list itself |
| `list.count(x)` | how many times `x` appears |
| `list.index(x)` | position of the first `x` |

````python
scores = [85, 92, 78, 90, 66]
print("Count:", len(scores))
print("Total:", sum(scores))
print("Average:", sum(scores) / len(scores))
print("Lowest:", min(scores))
print("Highest:", max(scores))
print("Sorted:", sorted(scores))
print("High to low:", sorted(scores, reverse=True))
print("Original:", scores)   # sorted() didn't change it
````

````python
names = ["Sara", "Ali", "Omar", "Ali"]
names.sort()                # changes the list itself
print(names)
print(names.count("Ali"))
print(names.index("Omar"))
````

## 7. Lists from User Input

### Option 1: `split()` a line of text

````python
text = input("Enter numbers separated by spaces: ")
numbers = []
for piece in text.split():
    numbers.append(int(piece))
print("Numbers:", numbers)
print("Sum:", sum(numbers))
````

### Option 2: keep asking with a loop

````python
fruits = []
while True:
    fruit = input("Add a fruit (or 'done'): ")
    if fruit == "done":
        break
    fruits.append(fruit)
print("Your fruits:", fruits)
````

## 8. Lists Inside Lists (2D lists)

A list can contain other lists — great for **tables**:

````python
grades = [
    ["Ali", 85],
    ["Sara", 92],
    ["Omar", 78],
]
print(grades[1])       # ['Sara', 92]
print(grades[1][0])    # Sara
for name, score in grades:
    print(f"{name}: {score}")
````

## 9. 🛠️ Project: To-Do List App

This combines **everything** so far: lists, `while` loops, `if/elif/else`, input and f-strings.

````python
todo = []

while True:
    print("\n===== TO-DO LIST =====")
    print("1. Add a task")
    print("2. Show tasks")
    print("3. Remove a task")
    print("4. Quit")
    choice = input("Choose an option (1-4): ")

    if choice == "1":
        task = input("Task to add: ")
        todo.append(task)
        print(f"✅ Added '{task}'")
    elif choice == "2":
        if len(todo) == 0:
            print("🎉 Nothing to do!")
        else:
            for number, task in enumerate(todo, start=1):
                print(f"{number}. {task}")
    elif choice == "3":
        number = int(input("Task number to remove: "))
        if 1 <= number <= len(todo):
            removed = todo.pop(number - 1)
            print(f"🗑️ Removed '{removed}'")
        else:
            print("❌ No task with that number")
    elif choice == "4":
        print("Goodbye! 👋")
        break
    else:
        print("❌ Please choose 1, 2, 3 or 4")
````
