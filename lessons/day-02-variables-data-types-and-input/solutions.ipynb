### Exercise 7 ⭐⭐ — Rectangle
Ask for the `length` and `width` of a rectangle (they may be decimals). Print the **area** (length × width) and the **perimeter** (2 × (length + width)).

### Exercise 8 ⭐⭐ — Sharing sweets 🍬
There are `47` sweets and `5` children. Using `//` and `%`, print how many sweets **each child** gets and how many are **left over**.

````python
sweets = 47
children = 5
print("Each child gets:", sweets // children)
print("Left over:", sweets % children)
````

### Exercise 9 ⭐⭐⭐ — Minutes converter
Ask the user for a number of minutes (e.g. `135`) and print it as hours and minutes: `135 minutes = 2 hours and 15 minutes`.

````python
total_minutes = int(input("Enter minutes: "))
hours = total_minutes // 60
minutes = total_minutes % 60
print(f"{total_minutes} minutes = {hours} hours and {minutes} minutes")
````

### Exercise 10 ⭐⭐⭐ — Mini Project: Shopping receipt 🧾
Ask the user for an item name, its price and the quantity. Print a receipt showing the item, the subtotal, a 6% tax and the total.

Example:
```
===== RECEIPT =====
Item:     Notebook
Price:    4.5 x 3
Subtotal: 13.5
Tax (6%): 0.81
Total:    14.31
===================
```
*(Hint: `round(number, 2)` rounds to 2 decimal places.)*

````python
item = input("Item name: ")
price = float(input("Price: "))
quantity = int(input("Quantity: "))

subtotal = price * quantity
tax = subtotal * 0.06
total = subtotal + tax

print("===== RECEIPT =====")
print("Item:    ", item)
print("Price:   ", price, "x", quantity)
print("Subtotal:", round(subtotal, 2))
print("Tax (6%):", round(tax, 2))
print("Total:   ", round(total, 2))
print("===================")
````
