### Exercise 4 ⭐⭐ — Vowel counter
Ask the user for a word and count how many vowels (a, e, i, o, u) it has. Capital letters count too!

````python
word = input("Enter a word: ").lower()
vowels = 0
for letter in word:
    if letter in "aeiou":
        vowels += 1
print("Vowels:", vowels)
````

### Exercise 5 ⭐⭐ — Email checker
Ask for an email address. Print `Valid email` if it contains `@`, contains `.`, and does not start with `@`. Otherwise print `Invalid email`.

````python
email = input("Enter your email: ").strip()
if "@" in email and "." in email and not email.startswith("@"):
    print("Valid email")
else:
    print("Invalid email")
````

### Exercise 6 ⭐⭐ — Username maker
Ask for the user's full name (e.g. `homaira mohammadi`) and make a username: the first name in lowercase + the first letter of the last name + the length of the full name. Example: `homairam17`.

````python
full_name = input("Enter your full name: ").strip().lower()
parts = full_name.split()
username = parts[0] + parts[-1][0] + str(len(full_name))
print("Your username is:", username)
````

### Exercise 9 ⭐⭐⭐ — Password strength checker 🔐
Ask the user to create a password. Keep asking until it is **strong**:
- at least 8 characters
- has at least one digit
- has at least one uppercase letter

Tell the user what is missing each time.
*(Hint: loop through the characters and use `isdigit()` and `isupper()`.)*

````python
while True:
    password = input("Create a password: ")
    has_digit = False
    has_upper = False
    for ch in password:
        if ch.isdigit():
            has_digit = True
        if ch.isupper():
            has_upper = True

    if len(password) < 8:
        print("❌ Too short (need at least 8 characters)")
    elif not has_digit:
        print("❌ Add at least one number")
    elif not has_upper:
        print("❌ Add at least one uppercase letter")
    else:
        print("✅ Strong password!")
        break
````

### Exercise 10 ⭐⭐⭐ — Mini Project: Secret message (Caesar cipher) 🕵️
Encrypt a message by moving every letter 3 places forward in the alphabet (`a`→`d`, `x`→`a`). Keep spaces and punctuation as they are.

*(Hint: `alphabet = "abcdefghijklmnopqrstuvwxyz"`, use `alphabet.find(letter)` and `% 26`.)*

````python
alphabet = "abcdefghijklmnopqrstuvwxyz"
message = "meet me at the park!"
shift = 3

secret = ""
for letter in message:
    if letter in alphabet:
        position = alphabet.find(letter)
        new_position = (position + shift) % 26
        secret += alphabet[new_position]
    else:
        secret += letter

print("Original: ", message)
print("Encrypted:", secret)
````
