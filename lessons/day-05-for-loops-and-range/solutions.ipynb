### Exercise 4 ⭐⭐ — Sum of even numbers
Calculate and print the sum of all **even** numbers from 1 to 100.

````python
total = 0
for n in range(2, 101, 2):
    total += n
print("Sum of even numbers 1-100:", total)
````

### Exercise 5 ⭐⭐ — Spell it out
Ask the user for their name and print each letter on its own line, with its position: `1: A`, `2: l`, `3: i`.

````python
name = input("Enter your name: ")
position = 1
for letter in name:
    print(f"{position}: {letter}")
    position += 1
````

### Exercise 6 ⭐⭐ — Class average
Ask how many students there are. Then ask for each student's score and print the class average and the highest score.

````python
how_many = int(input("How many students? "))
total = 0
highest = 0
for i in range(1, how_many + 1):
    score = int(input(f"Score of student {i}: "))
    total += score
    if score > highest:
        highest = score
print("Class average:", round(total / how_many, 1))
print("Highest score:", highest)
````

### Exercise 7 ⭐⭐ — Factorial
The factorial of 5 (written 5!) is 5 × 4 × 3 × 2 × 1 = 120. Write a loop that calculates `10!`.
*(Hint: start with `result = 1`, not 0!)*

````python
result = 1
for n in range(1, 11):
    result *= n
print("10! =", result)
````

### Exercise 10 ⭐⭐⭐ — Mini Project: Prime number finder 🔍
A **prime** number can only be divided evenly by 1 and itself (2, 3, 5, 7, 11, ...). Print all the prime numbers from 2 to 50.
*(Hint: for each number, use an inner loop to check if any smaller number divides it evenly.)*

````python
for number in range(2, 51):
    is_prime = True
    for divisor in range(2, number):
        if number % divisor == 0:
            is_prime = False
            break
    if is_prime:
        print(number, end=" ")
````
