## 🔁 Review of Day 4
- `while` repeats code **while** a condition is True
- Don't forget to **update** the counter, or you get an infinite loop!
- `break` leaves the loop, `continue` skips to the next round

Here is a Day 4 loop that prints 1 to 5:

````python
count = 1
while count <= 5:
    print(count)
    count += 1
````

That needed **3 lines** to manage the counter. A `for` loop does it in **one**:

````python
for count in range(1, 6):
    print(count)
````

---
## 1. The `for` Loop

```python
for variable in sequence:
    statement
```

A `for` loop goes through a **sequence** (a series of items) and runs the block **once for each item**.
Each time, the **loop variable** holds the current item.

- No counter to create ✅
- No counter to update ✅
- No infinite loops ✅

---
## 2. `range()` — Making a Sequence of Numbers

| Code | Numbers produced | Rule |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | starts at **0**, stops **before** 5 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | starts at 1, stops **before** 6 |
| `range(0, 11, 2)` | 0, 2, 4, 6, 8, 10 | third number = **step** |
| `range(10, 0, -1)` | 10, 9, ..., 1 | a negative step counts **down** |

> ⚠️ The most common surprise: the **stop number is NOT included**. `range(1, 5)` gives 1, 2, 3, 4.

````python
for i in range(5):
    print(i)
````

````python
for i in range(1, 6):
    print("Round", i)
````

````python
# Step: count by 2s
for i in range(0, 11, 2):
    print(i)
````

````python
# Counting down
for i in range(10, 0, -1):
    print(i)
print("Blast off! 🚀")
````

### Repeat something N times
When you don't need the number, programmers often call the loop variable `_`:

````python
for _ in range(3):
    print("Hip hip hooray! 🎉")
````

### Let the user choose

````python
stars = int(input("How many stars? "))
for i in range(stars):
    print("⭐")
````

## 3. Looping Through a String

A string is a **sequence of characters**, so a `for` loop can go through it letter by letter:

````python
for letter in "Python":
    print(letter)
````

````python
# Count the vowels in a word
word = "programming"
vowels = 0
for letter in word:
    if letter in "aeiou":
        vowels += 1
print(word, "has", vowels, "vowels")
````

> 💡 `letter in "aeiou"` is True if `letter` is one of those characters. We'll use `in` a lot on Days 6 and 7.

## 4. The Accumulator Pattern ➕

To add things up in a loop:
1. Create a variable **before** the loop (start at 0)
2. **Add** to it inside the loop
3. Use it **after** the loop

````python
total = 0
for number in range(1, 101):
    total += number
print("1 + 2 + ... + 100 =", total)
````

````python
# Average of numbers typed by the user
how_many = int(input("How many numbers? "))
total = 0
for i in range(how_many):
    number = float(input("Enter number: "))
    total += number
print("Average:", total / how_many)
````

### Loops + `if` together

````python
for number in range(1, 16):
    if number % 15 == 0:
        print("FizzBuzz")
    elif number % 3 == 0:
        print("Fizz")
    elif number % 5 == 0:
        print("Buzz")
    else:
        print(number)
````

> 👆 This is **FizzBuzz** — a famous programming puzzle! Multiples of 3 → Fizz, multiples of 5 → Buzz, both → FizzBuzz.

### Multiplication table

````python
number = 7
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
````

## 5. `break` and `continue` in `for` Loops

They work exactly like in `while` loops:

````python
# Find the first number divisible by 7 between 50 and 100
for n in range(50, 101):
    if n % 7 == 0:
        print("Found it:", n)
        break
````

````python
# Print 1-10 but skip the even numbers
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(n)
````

## 6. Nested Loops — A Loop Inside a Loop

The **inner** loop runs completely for **each** round of the outer loop.

````python
for row in range(1, 4):
    for col in range(1, 4):
        print(f"({row},{col})", end=" ")
    print()   # new line after each row
````

> 💡 `print(..., end=" ")` prints a space instead of moving to a new line. That's how we keep things on the same line.

### Drawing patterns 🎨

````python
# A right triangle
for row in range(1, 6):
    print("*" * row)
````

> 💡 `"*" * 3` gives `"***"` — multiplying a string repeats it!

````python
# A pyramid
height = int(input("Height of the pyramid: "))
for row in range(1, height + 1):
    spaces = " " * (height - row)
    stars = "*" * (2 * row - 1)
    print(spaces + stars)
````

````python
# Full multiplication grid (1-5)
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end="\t")   # \t = tab, lines up the columns
    print()
````

## 7. `for` or `while`?

| Use `for` when... | Use `while` when... |
|---|---|
| you know **how many times** to repeat | you repeat **until something happens** |
| you go through **each item** of a sequence | you don't know how many times |
| e.g. print 10 times, go through each letter | e.g. keep asking until the password is right |
