## 1. Why Functions?

A **function** is a named block of code that does **one job**. You write it once and **use it many times**.

- ♻️ **Reuse** — no copy-pasting the same code
- 📖 **Readability** — `calculate_average(scores)` explains itself
- 🔧 **Easier to fix** — fix a bug in **one** place
- 🧱 **Organise** — split a big problem into small pieces

Think of a function like a **recipe** 📜: you write the recipe once, then "cook" it whenever you want.

---
## 2. Defining and Calling a Function

```python
def function_name():
    # the function's code (indented!)
```

- `def` = **def**ine
- name in `snake_case` + brackets `()` + colon `:`
- the code inside is **indented**

Defining a function **does not run it**. You must **call** it by writing its name with `()`:

````python
def say_hello():
    print("Hello!")
    print("Welcome to Python 🐍")

# Nothing has happened yet... now we CALL it:
say_hello()
say_hello()
````

> ⚠️ You must **define** a function **before** you call it. Python reads from top to bottom!

---
## 3. Parameters — Giving a Function Information

**Parameters** are variables in the brackets. The values you pass when calling are called **arguments**.

````python
def greet(name):
    print(f"Hello, {name}! 👋")

greet("Ali")
greet("Sara")
greet("Homaira")
````

### More than one parameter

````python
def introduce(name, age, city):
    print(f"My name is {name}. I am {age} and I live in {city}.")

introduce("Ali", 20, "Kabul")
introduce("Sara", 19, "Kuala Lumpur")
````

> ⚠️ Order matters! The 1st argument goes to the 1st parameter, and so on. You can also use names to be clear:

````python
introduce(city="Tokyo", name="Omar", age=22)
````

### Default values
A parameter can have a **default** value, used when no argument is given:

````python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")

greet("Ali")                 # uses the default
greet("Sara", "Good morning")
greet("Omar", "Salaam")
````

## 4. `return` — Getting a Result Back

So far our functions only **printed**. Usually we want a function to **calculate** something and **give back** the result, so we can store it or use it later.

````python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)
print(add(10, 20) * 2)    # use the result in a calculation
````

### `print` vs `return` — a very common confusion!

| `print()` | `return` |
|---|---|
| **shows** a value on the screen | **gives back** a value to the code that called it |
| the value can't be used later | the value can be stored, used, printed... |
| the function continues after it | the function **stops** immediately |

````python
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b

x = add_print(2, 3)    # shows 5, but gives back nothing
y = add_return(2, 3)   # shows nothing, gives back 5
print("x is", x)       # None = "nothing"
print("y is", y)
````

### Functions that return True or False

````python
def is_even(number):
    return number % 2 == 0

print(is_even(4))
print(is_even(7))

for n in range(1, 6):
    if is_even(n):
        print(n, "is even")
````

### `return` stops the function

````python
def grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"   # only reached if nothing above returned

for s in [95, 72, 55, 30]:
    print(s, "→", grade(s))
````

### Functions with lists

````python
def average(numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

print(average([80, 90, 70]))
print(average([]))
````

## 5. Docstrings — Explain Your Function 📝

A **docstring** is a string right under `def` that explains what the function does. Good habit!

````python
def area_of_rectangle(length, width):
    """Return the area of a rectangle."""
    return length * width

print(area_of_rectangle(5, 3))
help(area_of_rectangle)
````

## 6. Local and Global Variables 🏠

Variables created **inside** a function are **local**: they only exist inside that function.

````python
def make_message():
    message = "I live inside the function"
    print(message)

make_message()
print(message)   # ❌ error: message doesn't exist out here
````

Variables created **outside** functions are **global**: functions can **read** them. But it's better to pass values in as **parameters** and get results with **return** — that keeps functions independent and easy to test.

````python
tax_rate = 0.06   # global

def price_with_tax(price):
    return price + price * tax_rate

print(price_with_tax(100))
````

## 7. Functions Calling Functions
Small functions can be combined to build bigger ones:

````python
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

def describe_weather(c):
    f = celsius_to_fahrenheit(c)
    if c >= 30:
        feeling = "hot 🥵"
    elif c >= 20:
        feeling = "nice 😎"
    else:
        feeling = "cold 🥶"
    return f"{c}°C ({f:.0f}°F) is {feeling}"

print(describe_weather(35))
print(describe_weather(24))
print(describe_weather(8))
````

````python
temp = float(input("Temperature in Celsius: "))
print(describe_weather(temp))
````

## 8. 🛠️ Project: Calculator with Functions

Each job gets its own function. The `main()` function runs the menu.

````python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: can't divide by zero"
    return a / b

def show_menu():
    print("\n1. Add  2. Subtract  3. Multiply  4. Divide  5. Quit")

def main():
    while True:
        show_menu()
        choice = input("Choose (1-5): ")
        if choice == "5":
            print("Bye! 👋")
            break
        if choice not in ["1", "2", "3", "4"]:
            print("Invalid choice")
            continue
        a = float(input("First value: "))
        b = float(input("Second value: "))
        if choice == "1":
            print("Result:", add(a, b))
        elif choice == "2":
            print("Result:", subtract(a, b))
        elif choice == "3":
            print("Result:", multiply(a, b))
        else:
            print("Result:", divide(a, b))

main()
````
