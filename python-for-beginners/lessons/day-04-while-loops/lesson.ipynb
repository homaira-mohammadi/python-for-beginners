## 1. Why Do We Need Loops?

Imagine you need to print "Hello" 5 times. Without a loop:

````python
print("Hello")
print("Hello")
print("Hello")
print("Hello")
print("Hello")
````

That works... but what about 100 times? Or 1,000? 😩

A **loop** repeats code for us. Programmers follow the **DRY** rule: **D**on't **R**epeat **Y**ourself.

Python has two kinds of loops:
| Loop | Use it when... |
|---|---|
| `while` | you repeat **while a condition is True** (you may not know how many times) |
| `for` | you repeat a **fixed number of times** or go through items (Day 5) |

---
## 2. The `while` Loop

```python
while condition:
    statement     # repeats as long as the condition is True
```

It looks just like an `if` — but instead of running **once**, it keeps running **again and again** while the condition is True.

```
   +------> < Is the condition True? > --No--> continue after the loop
   |                  |
   |                 Yes
   |                  v
   +------- Run the indented block
```

### Counting with a `while` loop

````python
count = 1
while count <= 5:
    print("Hello", count)
    count = count + 1   # VERY important: move towards the end!

print("Done!")
````

**Step by step:**

| `count` | `count <= 5`? | What happens |
|---|---|---|
| 1 | True | prints `Hello 1`, count becomes 2 |
| 2 | True | prints `Hello 2`, count becomes 3 |
| ... | ... | ... |
| 5 | True | prints `Hello 5`, count becomes 6 |
| 6 | **False** | loop stops → prints `Done!` |

Every loop has **3 parts**:
1. **Start** — `count = 1`
2. **Condition** — `while count <= 5:`
3. **Update** — `count = count + 1` (or `count += 1`)

### More examples

````python
# Countdown 🚀
countdown = 5
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Blast off! 🚀")
````

````python
# Even numbers from 2 to 10
number = 2
while number <= 10:
    print(number)
    number += 2
````

````python
# Add up the numbers from 1 to 100
total = 0
n = 1
while n <= 100:
    total += n
    n += 1
print("The sum of 1 to 100 is", total)
````

## 3. ⚠️ Infinite Loops

If the condition **never becomes False**, the loop runs **forever**. This is called an **infinite loop**.

```python
count = 1
while count <= 5:
    print("Hello")
    # oops! we forgot count += 1, so count is ALWAYS 1
```

> 🛑 **How to stop it:** In Jupyter / Colab press the **■ Stop** button (or *Interrupt*). In a terminal press **Ctrl + C**.
>
> ✅ **How to avoid it:** always check that something inside the loop changes so the condition will become False.

## 4. Loops with User Input

`while` loops are perfect when we **don't know** how many times to repeat — for example, until the user types `quit`:

````python
word = ""
while word != "quit":
    word = input("Type a word (or 'quit' to stop): ")
    print("You typed:", word)
print("Goodbye! 👋")
````

### Running total

````python
total = 0
number = int(input("Add (0 to stop): "))
while number != 0:
    total += number
    number = int(input("Add (0 to stop): "))
print("Total:", total)
````

### Input validation — keep asking until the answer is valid ✅

````python
rating = int(input("Rate this lesson between 1 and 5: "))
while rating < 1 or rating > 5:
    print("❌ Please enter a number between 1 and 5!")
    rating = int(input("Rate this lesson between 1 and 5: "))
print("Thanks for rating", rating, "⭐")
````

## 5. `break` — Leave the Loop Now

`break` **immediately stops** the loop, even if the condition is still True.
A common pattern is `while True:` (loop forever) + `break` (when we're done):

````python
while True:
    password = input("Password: ")
    if password == "python123":
        print("✅ Access granted!")
        break
    print("❌ Wrong password, try again.")
````

### Limiting the number of tries

````python
attempts = 0
while attempts < 3:
    pin = input("Enter your PIN: ")
    attempts += 1
    if pin == "2222":
        print("✅ Correct PIN!")
        break
    else:
        print("❌ Wrong PIN. Attempts used:", attempts)

if attempts == 3 and pin != "2222":
    print("🔒 Card blocked.")
````

---
## 6. `continue` — Skip to the Next Round

`continue` skips the **rest of this round** and jumps back to the condition.

````python
# print 1 to 10, but skip 5
n = 0
while n < 10:
    n += 1
    if n == 5:
        continue      # skip the print for 5
    print(n)
````

> ⚠️ With `continue`, make sure the update (`n += 1`) happens **before** it — otherwise you get an infinite loop!

## 7. 🎮 Game Upgrade: Guess the Number (with a loop!)

Remember Day 3's game with only **one** guess? Now the player keeps guessing until they win, and we count the tries.

````python
import random

secret = random.randint(1, 20)
tries = 0
guess = 0

print("I'm thinking of a number between 1 and 20...")
while guess != secret:
    guess = int(input("Your guess (1-20): "))
    tries += 1
    if guess < secret:
        print("Too low! ⬆️")
    elif guess > secret:
        print("Too high! ⬇️")

print("🎉 Correct! The number was", secret)
print("You needed", tries, "tries.")
````
