### Exercise 1 ⭐ — Count to 10
Use a `while` loop to print the numbers from 1 to 10.

```python
n = 1
while n <= 10:
    print(n)
    n += 1
```

### Exercise 2 ⭐ — Countdown
Print a countdown from 10 to 1, then print `Happy New Year! 🎆`.

```python
n = 10
while n >= 1:
    print(n)
    n -= 1
print("Happy New Year! 🎆")
```

### Exercise 3 ⭐ — Odd numbers
Print all odd numbers from 1 to 19.

```python
n = 1
while n <= 19:
    print(n)
    n += 2
```

### Exercise 4 ⭐⭐ — Times table
Ask the user for a number and print its times table from 1 to 10, like `7 x 1 = 7`.

```python
number = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(number, "x", i, "=", number * i)
    i += 1
```

### Exercise 5 ⭐⭐ — Adding machine
Keep asking the user for numbers. When they type `done`, stop and print the total.
*(Hint: check for `"done"` **before** converting with `int()`.)*

```python
total = 0
while True:
    entry = input("Number to add (or 'done'): ")
    if entry == "done":
        break
    total += int(entry)
print("Total:", total)
```

### Exercise 6 ⭐⭐ — Secret word
Keep asking the user to guess the secret word `python` until they get it right. Then print how many guesses they needed.

```python
guesses = 0
word = ""
while word != "python":
    word = input("Guess the secret word: ")
    guesses += 1
print("You got it in", guesses, "guesses!")
```

### Exercise 7 ⭐⭐ — Skip multiples of 3
Print the numbers from 1 to 15, but **skip** every multiple of 3 using `continue`.

```python
n = 0
while n < 15:
    n += 1
    if n % 3 == 0:
        continue
    print(n)
```

### Exercise 8 ⭐⭐ — Savings goal 💰
You save RM 50 every month. Use a loop to find out how many months it takes to save at least RM 1,000.

```python
savings = 0
months = 0
while savings < 1000:
    savings += 50
    months += 1
print("It takes", months, "months to save RM", savings)
```

### Exercise 9 ⭐⭐⭐ — Mini Project: Piggy bank 🐷
Start with a balance of 0. Keep asking the user how much to deposit. Stop when they enter `0`. After each deposit, print the new balance. Negative amounts are not allowed — print a warning and ignore them.

```python
balance = 0
while True:
    amount = float(input("How much to deposit? (0 to stop): "))
    if amount == 0:
        break
    if amount < 0:
        print("⚠️ You can't deposit a negative amount!")
        continue
    balance += amount
    print("New balance: RM", balance)
print("Final balance: RM", balance)
```

### Exercise 10 ⭐⭐⭐ — Guessing game with a limit
Upgrade the guessing game: the player gets only **5 tries**. If they don't guess it, print `Game over! The number was ...`.

```python
import random

secret = random.randint(1, 20)
tries = 0
won = False

while tries < 5:
    guess = int(input("Your guess (1-20): "))
    tries += 1
    if guess == secret:
        won = True
        break
    elif guess < secret:
        print("Too low!")
    else:
        print("Too high!")

if won:
    print("🎉 You won in", tries, "tries!")
else:
    print("Game over! The number was", secret)
```
