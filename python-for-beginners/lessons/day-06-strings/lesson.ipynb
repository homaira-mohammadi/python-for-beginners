## 1. Strings are Sequences of Characters

Every character in a string has a **position number** called an **index**. Python starts counting at **0**!

```
 P   y   t   h   o   n
 0   1   2   3   4   5     ← index
-6  -5  -4  -3  -2  -1     ← negative index (counts from the end)
```

````python
word = "Python"
print(word[0])    # first character
print(word[1])
print(word[5])    # last character
print(word[-1])   # last character (easy way!)
print(word[-2])   # second from the end
````

### `len()` — how long is a string?

````python
word = "Python"
print(len(word))

sentence = "I love coding"
print(len(sentence))   # spaces count too!
````

> ⚠️ If an index is too big, you get an `IndexError`:

````python
word = "Python"
print(word[10])
````

## 2. Slicing — Cutting Out Part of a String ✂️

```python
text[start:stop]       # from start up to (but NOT including) stop
text[start:stop:step]  # with a step
```

Just like `range()`, the **stop** position is not included.

````python
word = "Python"
print(word[0:3])   # Pyt
print(word[2:5])   # tho
print(word[:4])    # from the start: Pyth
print(word[2:])    # to the end: thon
print(word[-3:])   # last 3: hon
print(word[::2])   # every 2nd letter: Pto
print(word[::-1])  # reversed! nohtyP
````

## 3. Strings Cannot Be Changed (Immutable)

You can't change one character of a string directly:

````python
word = "Python"
word[0] = "J"
````

Instead, build a **new** string:

````python
word = "Python"
new_word = "J" + word[1:]
print(new_word)
````

## 4. String Methods 🛠️

A **method** is a function that belongs to a value. You use it with a **dot**: `text.upper()`.

### Changing case

````python
text = "hello World"
print(text.upper())      # HELLO WORLD
print(text.lower())      # hello world
print(text.title())      # Hello World
print(text.capitalize()) # Hello world
print(text)              # the original is NOT changed!
````

> 💡 Methods give back a **new** string. To keep the result, store it: `text = text.upper()`.

### Removing extra spaces — `strip()`
Users often type extra spaces by accident. `strip()` removes spaces from the **start and end**:

````python
name = "   Ali   "
print("[" + name + "]")
print("[" + name.strip() + "]")
````

### Replacing text — `replace()`

````python
sentence = "I like cats. Cats are cute."
print(sentence.replace("cats", "dogs"))
print(sentence.replace("Cats", "Dogs").replace("cats", "dogs"))
````

### Finding and counting — `find()` and `count()`

````python
sentence = "banana"
print(sentence.count("a"))    # how many times "a" appears: 3
print(sentence.find("n"))     # index of the FIRST "n": 2
print(sentence.find("z"))     # -1 means "not found"
````

### Checking — `startswith()`, `endswith()`, `isdigit()`, `isalpha()`
These methods answer **True** or **False**:

````python
filename = "homework.py"
print(filename.endswith(".py"))     # True
print(filename.startswith("home"))  # True

print("2026".isdigit())    # True: only digits
print("20a6".isdigit())    # False
print("Python".isalpha())  # True: only letters
````

> 💡 `isdigit()` is great for checking user input **before** using `int()`!

### Splitting and joining — `split()` and `join()`
`split()` breaks a string into pieces (a **list** — more about lists on Day 7!):

````python
sentence = "Python is really fun"
words = sentence.split()       # split on spaces
print(words)
print("Number of words:", len(words))

date = "24/10/2026"
print(date.split("/"))         # split on "/"

print("-".join(["2026", "10", "24"]))   # join pieces with "-"
````

## 5. Checking with `in`

`in` checks if some text appears **inside** a string:

````python
email = "homaira@example.com"
print("@" in email)       # True
print("gmail" in email)   # False

if "@" in email and "." in email:
    print("Looks like a valid email ✅")
````

## 6. Special Characters (Escape Sequences)

A backslash `\` gives the next character a special meaning:

| Code | Meaning |
|---|---|
| `\n` | new line |
| `\t` | tab (a big space) |
| `\"` | a double quote inside `" "` |
| `\\` | a real backslash |

````python
print("Line 1\nLine 2\nLine 3")
print("Name:\tAli")
print("She said \"Hello!\"")
print('It\'s easy')
````

### Multi-line strings
Three quotes let a string span many lines:

````python
poem = """Roses are red,
Violets are blue,
Python is awesome,
And so are you!"""
print(poem)
````

## 7. Formatting with f-strings ✨

You met f-strings on Day 2. They can do much more:

````python
name = "Sara"
price = 4.5
score = 0.8756

print(f"Hello {name}!")
print(f"Price: RM {price:.2f}")       # 2 decimal places → 4.50
print(f"Score: {score:.1%}")          # as a percentage → 87.6%
print(f"Shout: {name.upper()}")       # methods work inside { }
print(f"[{name:>10}]")                # right-align in 10 spaces
print(f"[{name:<10}]")                # left-align
print(f"[{name:^10}]")                # centre
````

### A neat table with alignment

````python
print(f"{'Item':<10}{'Price':>8}")
print("-" * 18)
print(f"{'Apple':<10}{1.5:>8.2f}")
print(f"{'Bread':<10}{3.25:>8.2f}")
print(f"{'Milk':<10}{4:>8.2f}")
````

## 8. Putting It Together 🧩

### Palindrome checker
A **palindrome** reads the same forwards and backwards: *level*, *racecar*, *madam*.

````python
word = input("Enter a word: ")
clean = word.lower()
if clean == clean[::-1]:
    print(f"✅ '{word}' is a palindrome!")
else:
    print(f"❌ '{word}' is not a palindrome.")
````

### Tidy up a name

````python
raw = input("Enter your full name: ")
name = raw.strip().title()
first = name.split()[0]
print(f"Hello, {name}!")
print(f"Your first name has {len(first)} letters.")
print(f"Your initials are {name[0]}.{name.split()[-1][0]}.")
````

### Word counter

````python
sentence = input("Enter a sentence: ")
print("Characters:", len(sentence))
print("Words:", len(sentence.split()))
print("'Python' appears", sentence.count("Python"), "times")
````
