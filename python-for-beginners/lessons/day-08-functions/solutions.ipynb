### Exercise 4 ⭐⭐ — Temperature converter
Write `to_fahrenheit(c)` and `to_celsius(f)`. Use them to convert 100°C and 98.6°F.
*(F = C × 9/5 + 32 · C = (F − 32) × 5/9)*

````python
def to_fahrenheit(c):
    return c * 9 / 5 + 32

def to_celsius(f):
    return (f - 32) * 5 / 9

print("100°C =", to_fahrenheit(100), "°F")
print("98.6°F =", round(to_celsius(98.6), 1), "°C")
````

### Exercise 5 ⭐⭐ — Biggest of three
Write `biggest(a, b, c)` that returns the largest of three numbers **without** using `max()`.

````python
def biggest(a, b, c):
    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c
    return largest

print(biggest(3, 9, 5))
print(biggest(10, 2, 7))
````

### Exercise 6 ⭐⭐ — Vowel counter function
Write `count_vowels(text)` that returns the number of vowels in `text`.

````python
def count_vowels(text):
    count = 0
    for letter in text.lower():
        if letter in "aeiou":
            count += 1
    return count

print(count_vowels("Programming is AWESOME"))
````

### Exercise 7 ⭐⭐ — Is it a palindrome?
Write `is_palindrome(word)` that returns `True` or `False`. Test it with `"Racecar"`, `"python"` and `"level"`.

````python
def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]

for w in ["Racecar", "python", "level"]:
    print(w, "→", is_palindrome(w))
````

### Exercise 8 ⭐⭐ — Discount price
Write `final_price(price, discount=10)` that returns the price after a percentage discount. If no discount is given, use 10%.

````python
def final_price(price, discount=10):
    return price - price * discount / 100

print(final_price(200))       # 10% off
print(final_price(200, 25))   # 25% off
````

### Exercise 9 ⭐⭐⭐ — List helpers
Write three functions that each take a list of numbers:
- `total(numbers)` — returns the sum (without `sum()`)
- `mean(numbers)` — returns the average (use `total()`!)
- `above_average(numbers)` — returns a **new list** of the numbers above the average

````python
def total(numbers):
    result = 0
    for n in numbers:
        result += n
    return result

def mean(numbers):
    return total(numbers) / len(numbers)

def above_average(numbers):
    avg = mean(numbers)
    result = []
    for n in numbers:
        if n > avg:
            result.append(n)
    return result

marks = [55, 80, 67, 92, 40, 75]
print("Total:", total(marks))
print("Mean:", round(mean(marks), 1))
print("Above average:", above_average(marks))
````

### Exercise 10 ⭐⭐⭐ — Mini Project: Report card 📄
Write `get_grade(score)` and `print_report(name, subjects, scores)`. `print_report` should print each subject's score with its grade, then the average and the overall grade.
```python
print_report("Sara", ["Maths", "Science", "English"], [88, 74, 65])
```
Expected output:
```
===== Report card: Sara =====
Maths       88  A
Science     74  B
English     65  C
----------------------
Average: 75.7  Overall grade: B
```

````python
def get_grade(score):
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    return "F"

def print_report(name, subjects, scores):
    print(f"===== Report card: {name} =====")
    for i in range(len(subjects)):
        print(f"{subjects[i]:<10} {scores[i]:>3}  {get_grade(scores[i])}")
    avg = sum(scores) / len(scores)
    print("-" * 22)
    print(f"Average: {avg:.1f}  Overall grade: {get_grade(avg)}")

print_report("Sara", ["Maths", "Science", "English"], [88, 74, 65])
````
