## 1. Conditions and Comparison Operators

A **condition** is a yes/no question. Python answers it with `True` or `False` (a `bool`!).

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `7 > 10` | `False` |
| `<` | Less than | `3 < 8` | `True` |
| `>=` | Greater than or equal | `18 >= 18` | `True` |
| `<=` | Less than or equal | `20 <= 15` | `False` |

> ⚠️ **`=` vs `==`** — one `=` **stores** a value (`age = 18`). Two `==` **compares** values (`age == 18`).

````python
age = 20
print(age >= 18)
print(age < 13)
print(age == 20)
print(age != 20)

name = "Alice"
print(name == "Alice")
print(name == "alice")   # False: capital letters matter!
````

## 2. The `if` Statement

An `if` statement tells Python: **"Run this code ONLY IF the condition is True."**

```python
if condition:
    statement
```

**Key parts:**
- `if` keyword
- a **condition** (True or False)
- a **colon** `:` at the end
- the code inside is **indented** (4 spaces)

### Your first `if` statement

````python
age = 20
if age >= 18:
    print("You are an adult.")
````

**Step by step:**
1. Store `20` in `age`
2. Check the condition: `20 >= 18` → **True**
3. Run the indented `print()`

### When the condition is False

````python
age = 15
if age >= 18:
    print("You are an adult.")
````

- Condition: `15 >= 18` → **False**
- Python **skips** the indented code — nothing prints
- The program simply continues

### How `if` works — flowchart

```
        [ Start ]
            |
    < Check condition >
     +------+------+
     |             |
   True          False
     |             |
  Run the       Skip the
  indented       block
  block            |
     |             |
     +------+------+
            |
 [ Continue with program ]
```

> ⚠️ **Key rules:** every `if` needs a colon `:` · the code inside must be indented (4 spaces) · without indentation Python doesn't know what belongs to the `if`.

### 🧠 Prediction challenge
What will this print? A) Nothing · B) "It is hot!" · C) "It is hot!" and "Have a nice day!" · D) "Have a nice day!"

````python
temperature = 35
if temperature > 30:
    print("It is hot!")
print("Have a nice day!")
````

**Answer: C.** `35 > 30` is True, so `"It is hot!"` prints. The second `print()` is **not indented**, so it is outside the `if` and **always** runs.

> **Indentation is everything in Python.** It defines which code is *inside* the `if`.

Try changing `temperature` to `20` and run it again — what changes?

### Using `if` with user input

````python
score = int(input("Enter your score: "))
if score >= 50:
    print("You passed!")
````

| Input | Condition | Output |
|---|---|---|
| 75 | True | You passed! |
| 40 | False | *(nothing)* |
| 50 | True | You passed! |

> ⚠️ Use `int()` to convert input to a number. `input()` always returns a string, and you can't compare a string with a number.

## 3. `else` — The Other Choice

`else` runs when the `if` condition is **False**. Think of it as: *"If this isn't true, do this instead."*

```python
if condition:
    statement      # runs when True
else:
    statement      # runs when False
```

```
     < Check condition >
             |
     +-------+-------+
     |               |
    True           False
     |               |
 Run the if      Run the else
   block            block
```

> ⚠️ `else` does **not** have a condition. It catches everything the `if` didn't.

````python
age = 15
if age >= 18:
    print("You are an adult.")
else:
    print("You are under 18.")
````

## 4. `elif` — Multiple Choices

Sometimes there are **more than two** possible outcomes. For example, grading a score:

| Score | 80–100 | 70–79 | 60–69 | 50–59 | Below 50 |
|---|---|---|---|---|---|
| Grade | A | B | C | D | F |

`elif` (short for **else if**) lets Python check **another** condition when the previous one was False.

```python
if condition1:
    statement      # checked first, always
elif condition2:
    statement      # checked only if condition1 was False
else:
    statement      # runs if ALL conditions were False
```

You can have as many `elif`s as you need. `else` is optional.

### Simple example

````python
score = 75
if score >= 80:
    print("Grade A")
elif score >= 70:
    print("Grade B")
else:
    print("Keep practicing!")
````

**Why "Grade B"?**
1. Check `if`: `75 >= 80` → False ❌ — skip
2. Check `elif`: `75 >= 70` → True ✅ — run this block
3. Python prints **Grade B** and **stops checking**.

### How Python checks `elif`

```
 < Condition 1? > --True--> Run block 1
        |
      False
        v
 < Condition 2? > --True--> Run block 2
        |
      False
        v
 < Condition 3? > --True--> Run block 3
        |
      False
        v
  Run else block
```

> Python checks **from top to bottom** and stops at the **first True** condition — even if later conditions would also be True. That's why **order matters**!

### Full grade checker

````python
score = 64
if score >= 80:
    print("Grade A")
elif score >= 70:
    print("Grade B")
elif score >= 60:
    print("Grade C")
elif score >= 50:
    print("Grade D")
else:
    print("Grade F")
````

### ⚠️ Why order matters
This version is **wrong**. A score of 95 gets "Grade D", because `95 >= 50` is checked first and is already True:

````python
score = 95
if score >= 50:
    print("Grade D")
elif score >= 80:
    print("Grade A")   # never reached for high scores!
````

## 5. Logical Operators: `and`, `or`, `not`

Sometimes one decision depends on **two or more** conditions.

| Operator | Meaning | Example | True when... |
|---|---|---|---|
| `and` | both must be True | `age >= 18 and has_ticket` | **both** are True |
| `or` | at least one must be True | `day == "Sat" or day == "Sun"` | **either one** is True |
| `not` | flips True ↔ False | `not is_raining` | `is_raining` is False |

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("Welcome to the concert! 🎵")
else:
    print("Sorry, you can't enter.")
````

````python
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend! 🎉")
else:
    print("It's a school day. 📚")
````

````python
is_raining = False

if not is_raining:
    print("Let's play outside! ⚽")
````

### Checking a range
`and` is perfect for checking if a number is **between** two values:

````python
temperature = 25
if temperature >= 20 and temperature <= 30:
    print("Perfect weather!")
````

### Login example (two conditions)

````python
username = input("Username: ")
password = input("Password: ")

if username == "admin" and password == "python123":
    print("✅ Login successful. Welcome!")
else:
    print("❌ Wrong username or password.")
````

## 6. Nested `if` (an `if` inside an `if`)

You can put an `if` **inside** another `if`. Notice the **extra indentation**:

````python
age = 20
has_id = False

if age >= 18:
    print("Age check passed.")
    if has_id:
        print("You can enter.")
    else:
        print("Please show your ID.")
else:
    print("Sorry, you are too young.")
````

## 7. 🎮 Let's Build a Game: Guess the Number

The computer picks a secret number from 1 to 10. You get **one guess**!

`random.randint(1, 10)` picks a random whole number from 1 to 10.
`import random` loads Python's built-in *random* tool so we can use it.

````python
import random

secret = random.randint(1, 10)
guess = int(input("Guess a number between 1 and 10: "))

if guess == secret:
    print("🎉 Amazing! You guessed it!")
elif guess < secret:
    print("Too low! The number was", secret)
else:
    print("Too high! The number was", secret)
````

> 💡 Only one guess feels short, right? On **Day 4** we'll learn **loops** so the player can keep guessing until they win!

### 🎮 Bonus game: Rock, Paper, Scissors ✊✋✌️

````python
import random

computer = random.choice(["rock", "paper", "scissors"])   # picks one option (lists come on Day 7!)
player = input("Choose rock, paper or scissors: ").lower()

print("Computer chose:", computer)

if player == computer:
    print("It's a tie! 🤝")
elif (player == "rock" and computer == "scissors") or \
     (player == "paper" and computer == "rock") or \
     (player == "scissors" and computer == "paper"):
    print("You win! 🏆")
elif player == "rock" or player == "paper" or player == "scissors":
    print("Computer wins! 🤖")
else:
    print("That's not a valid choice.")
````

> `.lower()` turns the user's answer into small letters, so `Rock` and `ROCK` work too. We'll learn more about this on Day 6.
