### Exercise 10 ⭐⭐⭐ — Simple calculator 🧮
Ask for two numbers and an operator (`+`, `-`, `*` or `/`). Print the result. If the operator is something else, print `Unknown operator`.
**Extra:** if the user tries to divide by 0, print `You can't divide by zero!`

````python
num1 = float(input("First number: "))
operator = input("Operator (+, -, *, /): ")
num2 = float(input("Second number: "))

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 == 0:
        print("You can't divide by zero!")
    else:
        print("Result:", num1 / num2)
else:
    print("Unknown operator")
````

### Exercise 11 ⭐⭐⭐ — Mini Project: BMI checker ⚖️
Ask for weight (kg) and height (m). Calculate BMI = weight ÷ (height × height) and print the BMI (rounded to 1 decimal) and the category:

| BMI | Category |
|---|---|
| below 18.5 | Underweight |
| 18.5 – 24.9 | Normal |
| 25 – 29.9 | Overweight |
| 30 and above | Obese |

````python
weight = float(input("Weight in kg: "))
height = float(input("Height in m: "))
bmi = weight / (height * height)

print("Your BMI is", round(bmi, 1))
if bmi < 18.5:
    print("Category: Underweight")
elif bmi < 25:
    print("Category: Normal")
elif bmi < 30:
    print("Category: Overweight")
else:
    print("Category: Obese")
````
