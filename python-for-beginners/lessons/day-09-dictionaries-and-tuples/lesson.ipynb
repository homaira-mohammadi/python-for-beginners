## 1. What is a Dictionary?

A **dictionary** stores pairs of **key → value**, like a real dictionary: **word → meaning**, or a phone book: **name → number**.

```python
my_dict = {key1: value1, key2: value2}
```

- curly brackets `{ }`
- each pair is `key: value`
- pairs are separated by commas
- keys are usually strings and must be **unique**

````python
student = {
    "name": "Sara",
    "age": 19,
    "major": "Computer Science",
    "gpa": 3.7,
}
print(student)
print(student["major"])   # much clearer than student[2]!
print(student["gpa"])
print(len(student))       # number of pairs
````

---
## 2. Getting Values Safely

Using a key that doesn't exist causes a `KeyError`:

````python
student = {"name": "Sara", "age": 19}
print(student["email"])
````

Two safe ways:
1. check with `in` first
2. use `.get()` — it returns `None` (or a default you choose) instead of crashing

````python
student = {"name": "Sara", "age": 19}

if "email" in student:
    print(student["email"])
else:
    print("No email saved")

print(student.get("email"))                 # None
print(student.get("email", "not given"))    # custom default
print(student.get("name"))
````

---
## 3. Adding, Changing and Removing ✏️

````python
student = {"name": "Sara", "age": 19}

student["email"] = "sara@example.com"   # ADD a new key
student["age"] = 20                      # CHANGE an existing key
print(student)

del student["email"]                     # REMOVE by key
print(student)

age = student.pop("age")                 # remove AND get the value
print("Removed age:", age)
print(student)
````

> 💡 Same syntax for adding and changing: `dict[key] = value`. If the key exists it's **changed**; if not, it's **added**.

## 4. Looping Through a Dictionary 🔁

````python
prices = {"apple": 1.5, "bread": 3.25, "milk": 4.0}

print("Keys:")
for item in prices:              # loops through the KEYS
    print(" ", item)

print("Values:")
for price in prices.values():
    print(" ", price)

print("Both:")
for item, price in prices.items():
    print(f"  {item}: RM {price:.2f}")
````

````python
# Total of all values
prices = {"apple": 1.5, "bread": 3.25, "milk": 4.0}
print("Total:", sum(prices.values()))
````

## 5. Practical Examples

### A mini English → Malay dictionary 📖

````python
translations = {
    "hello": "helo",
    "thank you": "terima kasih",
    "water": "air",
    "friend": "kawan",
}

for _ in range(2):
    word = input("Look up a word: ").lower()
    print("→", translations.get(word, "Sorry, I don't know that word yet."))
````

### Counting things — a word counter
A dictionary is perfect for counting: the **key** is the thing, the **value** is how many.

````python
sentence = input("Paste a sentence: ")
counts = {}
for word in sentence.lower().split():
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1
print(counts)
````

## 6. Lists of Dictionaries — Storing Records 📇

Each dictionary is **one record** (one student). A list holds **all** the records:

````python
students = [
    {"name": "Ali", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Omar", "score": 47},
]

for s in students:
    status = "passed ✅" if s["score"] >= 50 else "failed ❌"
    print(f"{s['name']:<6} {s['score']:>3}  {status}")

best = students[0]
for s in students:
    if s["score"] > best["score"]:
        best = s
print("Top student:", best["name"])
````

> 💡 `"passed" if condition else "failed"` is a one-line `if/else` (called a **conditional expression**).

### Dictionaries with lists inside

````python
student = {
    "name": "Mariam",
    "subjects": ["Maths", "Physics", "Python"],
}
student["subjects"].append("Art")
print(student["name"], "studies", len(student["subjects"]), "subjects:")
for subject in student["subjects"]:
    print(" -", subject)
````

## 7. Tuples — Lists That Can't Change 🔒

A **tuple** is like a list, but written with **round brackets** `( )` and it **cannot be changed** after it's created (it's *immutable*).

Use a tuple for data that should stay the same: coordinates, RGB colours, days of the week.

````python
point = (3, 5)
colour = (255, 0, 128)
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")

print(point[0])
print(days[-1])
print(len(days))
print("Fri" in days)
````

````python
days = ("Mon", "Tue", "Wed")
days[0] = "Monday"    # ❌ tuples can't be changed
````

### Tuple unpacking
You've already done this! `x, y = 10, 20` from Day 2 uses a tuple. You can unpack any tuple into variables:

````python
point = (3, 5)
x, y = point
print("x =", x, "and y =", y)

def min_max(numbers):
    return min(numbers), max(numbers)   # return TWO values as a tuple

low, high = min_max([4, 9, 1, 7])
print("Lowest:", low, "Highest:", high)
````

### List vs tuple vs dictionary

| | List | Tuple | Dictionary |
|---|---|---|---|
| Brackets | `[ ]` | `( )` | `{ }` |
| Get items by | position | position | key |
| Can change? | ✅ yes | ❌ no | ✅ yes |
| Example | `["Ali", "Sara"]` | `(3, 5)` | `{"name": "Ali"}` |
| Use for | a changing collection | fixed data | labelled data |

## 8. 🛠️ Project: Contact Book 📒

````python
contacts = {}

while True:
    print("\n=== CONTACT BOOK ===")
    print("1. Add/update  2. Search  3. Delete  4. Show all  5. Quit")
    option = input("Menu option: ")

    if option == "1":
        name = input("Contact name: ").title()
        phone = input("Phone number: ")
        contacts[name] = phone
        print(f"✅ Saved {name}")
    elif option == "2":
        name = input("Contact name: ").title()
        print(f"📞 {name}: {contacts.get(name, 'not found')}")
    elif option == "3":
        name = input("Contact name: ").title()
        if name in contacts:
            del contacts[name]
            print(f"🗑️ Deleted {name}")
        else:
            print("❌ Not found")
    elif option == "4":
        if not contacts:
            print("(empty)")
        for name, phone in sorted(contacts.items()):
            print(f"  {name:<10} {phone}")
    elif option == "5":
        print("Goodbye! 👋")
        break
    else:
        print("Please choose 1-5")
````
