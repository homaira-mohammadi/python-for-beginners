### Exercise 4 ⭐⭐ — Shopping bill 🛒
Given the prices and the cart below, calculate and print the total bill.
```python
prices = {"apple": 1.5, "banana": 0.5, "milk": 4.2, "bread": 3.0}
cart = {"apple": 4, "milk": 2, "bread": 1}   # item: quantity
```

````python
prices = {"apple": 1.5, "banana": 0.5, "milk": 4.2, "bread": 3.0}
cart = {"apple": 4, "milk": 2, "bread": 1}

total = 0
for item, quantity in cart.items():
    cost = prices[item] * quantity
    print(f"{item:<8} {quantity} x {prices[item]:.2f} = {cost:.2f}")
    total += cost
print(f"Total: RM {total:.2f}")
````

### Exercise 5 ⭐⭐ — Letter counter
Count how many times each letter appears in `"mississippi"` and print the dictionary.

````python
word = "mississippi"
counts = {}
for letter in word:
    counts[letter] = counts.get(letter, 0) + 1
print(counts)
````

### Exercise 6 ⭐⭐ — Grade book
Given `grades = {"Ali": 78, "Sara": 92, "Omar": 45, "Mariam": 88}`, print:
- every student who passed (50+)
- the class average
- the name of the top student

````python
grades = {"Ali": 78, "Sara": 92, "Omar": 45, "Mariam": 88}

print("Passed:")
for name, score in grades.items():
    if score >= 50:
        print(" -", name)

print("Average:", sum(grades.values()) / len(grades))

top = ""
top_score = -1
for name, score in grades.items():
    if score > top_score:
        top, top_score = name, score
print("Top student:", top)
````

### Exercise 7 ⭐⭐ — Tuples
Create a tuple of the 7 days of the week. Print the first and last day, and use a loop to print only the days that start with `"T"`.

````python
days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")
print("First:", days[0])
print("Last:", days[-1])
for day in days:
    if day.startswith("T"):
        print(day)
````

### Exercise 8 ⭐⭐⭐ — Library records 📚
Create a **list of dictionaries**, each with `title`, `author` and `year`. Add at least 4 books. Print the books published **after 2000**, and the title of the oldest book.

````python
books = [
    {"title": "Harry's Python", "author": "A. Khan", "year": 1999},
    {"title": "Coding for Kids", "author": "S. Lee", "year": 2015},
    {"title": "The Data Story", "author": "M. Noor", "year": 2008},
    {"title": "Old Computers", "author": "J. Smith", "year": 1985},
]

print("Published after 2000:")
for book in books:
    if book["year"] > 2000:
        print(f" - {book['title']} by {book['author']} ({book['year']})")

oldest = books[0]
for book in books:
    if book["year"] < oldest["year"]:
        oldest = book
print("Oldest book:", oldest["title"])
````

### Exercise 9 ⭐⭐⭐ — Mini Project: Quiz from a dictionary 🧠
Store quiz questions as a dictionary `{question: answer}`. Ask every question, keep score, and at the end print the score and a message (`Excellent!` for full marks, `Good job!` for at least half, otherwise `Keep practicing!`).

````python
quiz = {
    "What keyword defines a function? ": "def",
    "Which brackets make a list? ": "[]",
    "What does len('cat') return? ": "3",
    "Can a tuple be changed? (yes/no) ": "no",
}

score = 0
for question, answer in quiz.items():
    reply = input(question).strip().lower()
    if reply == answer:
        print("✅ Correct!")
        score += 1
    else:
        print("❌ Answer:", answer)

print(f"\nScore: {score}/{len(quiz)}")
if score == len(quiz):
    print("Excellent! 🏆")
elif score >= len(quiz) / 2:
    print("Good job! 👍")
else:
    print("Keep practicing! 💪")
````
