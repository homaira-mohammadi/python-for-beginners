## 1. Why Do We Need Variables?

Imagine storing a student's information **without** variables:

````python
print("Homaira")
print(20)
print("Computer Science")
````

**Problems:**
- ❌ Difficult to manage — what does `20` mean?
- ❌ Difficult to change — if the age changes, we must find and change it everywhere
- ❌ Repeating data — we have to type the same values again and again

Now **with** variables, the information is stored with a clear name and can be reused:

````python
name = "Homaira"
age = 20
major = "Computer Science"

print(name)
print(age)
print(major)
````

## 2. What is a Variable?

**A variable is a container that stores a value.**

Think of a **labelled box**: the label is the variable's *name*, and the thing inside is its *value*.

```
+-------------+
|    name     |   ← label (variable name)
|   "Ahmad"   |   ← what's inside (value)
+-------------+
```

In Python:

````python
name = "Ahmad"
print(name)
````

---
## 3. Creating Variables

**Syntax:**
```python
variable_name = value
```

### What does `=` mean?
In Python, `=` means **"assign (store) this value in this variable"**.
It does **NOT** mean "is equal to" like in maths. Read `age = 18` as *"age gets 18"*.

````python
name = "Ali"
age = 18
print(name, "is", age, "years old")
````

### Variable examples
Each variable below stores a different **type** of information — text, a whole number and a decimal:

````python
student_name = "Sara"
student_age = 19
student_gpa = 3.5

print(student_name)
print(student_age)
print(student_gpa)
````

> ⚠️ **`print(name)` vs `print("name")`** — without quotes Python prints the **value stored in the variable**; with quotes it prints the word itself.

````python
name = "Ali"
print(name)     # the value: Ali
print("name")   # the text: name
````

## 4. Variable Naming Rules

| Rule | ❌ Wrong | ✅ Correct |
|---|---|---|
| 1. Cannot start with a number | `1name = "Ali"` | `name1 = "Ali"` |
| 2. No spaces allowed (use `_`) | `student name = "Ali"` | `student_name = "Ali"` |
| 3. Only letters, numbers and `_` | `my-name = "Ali"` | `my_name = "Ali"` |
| 4. Don't use Python's keywords | `if = 5` | `if_value = 5` |

**Python is case-sensitive:** `Name`, `name` and `NAME` are three **different** variables.

````python
Name = "Big Ali"
name = "small ali"
NAME = "LOUD ALI"
print(Name)
print(name)
print(NAME)
````

Run this to see what happens when a rule is broken:

````python
1name = "Ali"
````

### Good variable names
Clear names make code easier to understand — for you and everyone who reads it later.

| Quality | Example |
|---|---|
| ❌ Bad | `x = 20` |
| ✅ Good | `student_age = 20` |

> 💡 Python programmers write names in **snake_case**: all lowercase, with `_` between words — `first_name`, `total_price`.

## 5. Multiple Assignment & Changing Values

### Multiple assignment
Python can assign several variables in one line:

````python
x, y, z = 10, 20, 30
print(x, y, z)
````

That is the same as writing:
```python
x = 10
y = 20
z = 30
```

### Changing variable values
Variables can be **updated** at any time — the old value is replaced:

````python
score = 50
print(score)

score = 80
print(score)
````

You can even use a variable's **old value** to calculate its new value:

````python
score = 80
score = score + 10   # take the old score (80), add 10, store it back
print(score)
````

> Variables can change — that's what makes them powerful! (That's why they are called *variables*.)

## 6. Introduction to Data Types

A **data type** tells Python **what kind of information** is stored.

| Type | Meaning | Example |
|---|---|---|
| `int` | Whole number (integer) | `10`, `-3`, `2026` |
| `float` | Decimal number | `3.5`, `0.99`, `-2.0` |
| `str` | Text (string) — always in quotes | `"Python"`, `'Ali'` |
| `bool` | True or False (boolean) | `True`, `False` |

### Checking the type with `type()`

````python
age = 20
gpa = 3.5
name = "Sara"
is_student = True

print(type(age))
print(type(gpa))
print(type(name))
print(type(is_student))
````

> ⚠️ `"20"` (with quotes) is a **string**, but `20` (no quotes) is an **int**. They look similar but behave very differently!

````python
print(type(20))
print(type("20"))
````

> 💡 `True` and `False` must start with a **capital letter** and have **no quotes**.

## 7. Arithmetic Operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `7 + 2` | `9` |
| `-` | Subtraction | `7 - 2` | `5` |
| `*` | Multiplication | `7 * 2` | `14` |
| `/` | Division | `7 / 2` | `3.5` |
| `//` | Floor division (whole part only) | `7 // 2` | `3` |
| `%` | Modulus (the **remainder**) | `7 % 2` | `1` |
| `**` | Power (exponent) | `7 ** 2` | `49` |

````python
a = 7
b = 2
print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** b =", a ** b)
````

### Why is `%` useful?
The remainder tells us if a number divides evenly. If `number % 2` is `0`, the number is **even**. We'll use this on Day 3!

````python
print(10 % 2)   # 0 -> 10 is even
print(7 % 2)    # 1 -> 7 is odd
````

### Order of operations
Python follows the maths rules: brackets first, then `**`, then `* / // %`, then `+ -`.

````python
print(2 + 3 * 4)     # 14, multiplication first
print((2 + 3) * 4)   # 20, brackets first
````

### Shortcut operators
`score += 10` is a short way to write `score = score + 10`. The same works for `-=`, `*=`, `/=`.

````python
coins = 5
coins += 3   # coins = coins + 3
print(coins)
coins -= 1   # coins = coins - 1
print(coins)
````

## 8. Getting Input from the User

`input()` lets the user **type** something. The program waits until they press **Enter**.
The text inside the brackets is the question (the *prompt*) shown to the user.

````python
name = input("What is your name? ")
print("Hello", name)
````

### ⚠️ Very important: `input()` always gives you a **string**
Even if the user types a number, Python stores it as text:

````python
age = input("Enter your age: ")
print(type(age))
````

So maths with it causes an error:

````python
age = input("Enter your age: ")
print(age + 5)
````

`TypeError: can only concatenate str (not "int") to str` → Python can't add text and a number.

## 9. Type Conversion

We can **convert** a value from one type to another:

| Function | Converts to | Example | Result |
|---|---|---|---|
| `int()` | whole number | `int("20")` | `20` |
| `float()` | decimal | `float("3.5")` | `3.5` |
| `str()` | text | `str(20)` | `"20"` |

Now the age example works:

````python
age = int(input("Enter your age: "))
print("Next year you will be", age + 1)
````

> 🧠 **Read it from the inside out:** `input()` gets the text `"20"` → `int()` turns it into the number `20` → it is stored in `age`.

Use `float()` when the user might type a decimal:

````python
price = float(input("Enter the price: "))
print("Price with 10% tax:", price * 1.1)
````

## 10. Combining Text and Variables

### Option 1: commas (easiest ✅)
Commas let you mix strings and numbers. Python adds spaces for you.

````python
name = "Ali"
age = 20
print("Hello", name)
print("You are", age, "years old")
````

### Option 2: `+` (only joins strings together)
`+` joins strings (this is called *concatenation*). It does **not** add spaces, and it does **not** work with numbers unless you convert them with `str()`:

````python
first = "Homaira"
last = "Mohammadi"
print(first + " " + last)
print("Age: " + str(20))
````

````python
print("Age: " + 20)   # ❌ error: can't join a string and an int
````

### Option 3: f-strings (the modern way ⭐)
Put an `f` before the quotes and write variables inside `{ }`:

````python
name = "Ali"
age = 20
print(f"My name is {name} and I am {age} years old.")
print(f"In 5 years I will be {age + 5}.")
````
