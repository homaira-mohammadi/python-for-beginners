### Exercise 2 ⭐ — Change the list
Start with `colours = ["red", "green", "blue"]`. Add `"yellow"` to the end, change `"green"` to `"purple"`, remove `"red"`, and print the list after each step.

````python
colours = ["red", "green", "blue"]
colours.append("yellow")
print(colours)
colours[1] = "purple"
print(colours)
colours.remove("red")
print(colours)
````

### Exercise 3 ⭐ — Loop and print
Given `cities = ["Kabul", "Kuala Lumpur", "Tokyo", "Paris"]`, print each city in UPPERCASE, numbered from 1.

````python
cities = ["Kabul", "Kuala Lumpur", "Tokyo", "Paris"]
for number, city in enumerate(cities, start=1):
    print(f"{number}. {city.upper()}")
````

### Exercise 4 ⭐⭐ — Score statistics
Keep asking for scores until the user types `done`. Then print the number of scores, the average, the highest and the lowest.

````python
scores = []
while True:
    entry = input("Enter a score (or 'done'): ")
    if entry == "done":
        break
    scores.append(int(entry))

print("Number of scores:", len(scores))
print("Average:", round(sum(scores) / len(scores), 1))
print("Highest:", max(scores))
print("Lowest:", min(scores))
````

### Exercise 5 ⭐⭐ — Passed students
Given the two lists below, print the names of students who scored **50 or more**.
```python
names = ["Ali", "Sara", "Omar", "Mariam", "Zain"]
scores = [45, 88, 67, 39, 72]
```
*(Hint: loop with `range(len(names))` and use the same index in both lists.)*

````python
names = ["Ali", "Sara", "Omar", "Mariam", "Zain"]
scores = [45, 88, 67, 39, 72]
for i in range(len(names)):
    if scores[i] >= 50:
        print(names[i], "passed with", scores[i])
````

### Exercise 6 ⭐⭐ — No duplicates
Keep asking for items to buy until `done`. Only add an item if it is **not already** in the list. Print the final shopping list.

````python
shopping = []
while True:
    item = input("Item to buy (or 'done'): ")
    if item == "done":
        break
    if item in shopping:
        print(f"'{item}' is already on the list!")
    else:
        shopping.append(item)
print("Shopping list:", shopping)
````

### Exercise 7 ⭐⭐ — Find the maximum yourself
Find the biggest number in `numbers = [34, 7, 89, 12, 56, 3]` **without** using `max()`.

````python
numbers = [34, 7, 89, 12, 56, 3]
biggest = numbers[0]
for n in numbers:
    if n > biggest:
        biggest = n
print("Biggest:", biggest)
````

### Exercise 8 ⭐⭐ — Long words
From `sentence = "Python is a powerful and friendly programming language"`, make a list of only the words with **more than 5 letters**.

````python
sentence = "Python is a powerful and friendly programming language"
long_words = []
for word in sentence.split():
    if len(word) > 5:
        long_words.append(word)
print(long_words)
````

### Exercise 9 ⭐⭐⭐ — Reverse without help
Reverse the list `[1, 2, 3, 4, 5]` into a **new** list **without** using `reverse()`, `reversed()` or `[::-1]`.

````python
numbers = [1, 2, 3, 4, 5]
backwards = []
for n in numbers:
    backwards.insert(0, n)
print(backwards)
````

### Exercise 10 ⭐⭐⭐ — Mini Project: Quiz with lists 🧠
Store 3 questions in one list and their answers in another list. Ask each question, check the answer (ignore capital letters and extra spaces), and print the final score.

````python
questions = [
    "What is the capital of Malaysia? ",
    "How many days are in a week? ",
    "Which language are we learning? ",
]
answers = ["kuala lumpur", "7", "python"]

score = 0
for i in range(len(questions)):
    reply = input(questions[i]).strip().lower()
    if reply == answers[i]:
        print("✅ Correct!")
        score += 1
    else:
        print("❌ The answer was:", answers[i])

print(f"You scored {score} out of {len(questions)}")
````
